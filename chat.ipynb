```python
from openai import OpenAI
import config
```

```python
client =OpenAI(api_key = config.api_key)
```

```python
def medical_chatbot(prompt):
    response = client.responses.create(
        model = "gpt-4.1-mini",
        input = [
            {
                "role" : "system",
                "content":"You are a helpful medical assistant chatbot.Provide possible causes based on symptoms.Ask 1-2 questions if needed.Suggest basic precautions.Advise immediate medical attention for severe symptoms.Never replace professional medical advice.Give in short and simple words"

            },
            {
                "role":"user",
                "content":prompt
            }
        ],
        max_output_tokens = 256
    )
    disclaimer = "\n\nDisclaimer: The information provided by this chatbot is for general informational purposes only and is not a substitute for professional medical advice, diagnosis, or treatment. Seek guidance from a licensed healthcare provider for any medical condition."

    return response.output_text + disclaimer
```

```python
prompt = "stomach pain with fever"
medical_chatbot(prompt)
```
